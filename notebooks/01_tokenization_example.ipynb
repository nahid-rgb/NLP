```python
corpus = """I love learning Natural Language Processing.
NLP is a very interesting field of Artificial Intelligence.
I am learning NLP with Python.
Python makes NLP practical and easy to understand.
Natural Language Processing helps computers understand human language.
I want to build useful NLP projects."""
```

```python
print(corpus)
```

```python
# Tokenization 
# Sentence --> paragraph 
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
```

```python

documents = sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize 
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
