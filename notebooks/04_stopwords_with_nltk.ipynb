# Stopwords

Stopwords are common words that usually don't carry much useful meaning for a particular NLP task.

Example:
```text
the, is, am, are, a, an, in, on, and, of, to, for
```
For example:
```text
"I am learning NLP in Python"
```
After removing stopwords:
```text
"learning NLP Python"
```

The words **I**, **m**, and **in** are removed because they often don't add much information for tasks like text classification or search.

Why remove stopwords?

It can:

- Reduce the number of words the model needs to process
- Reduce unnecessary/common words
- Make the text representation smaller
- Sometimes improve NLP/ML model performance

````python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
````

````python
import nltk
nltk.download('stopwords')
````

````python
paragraph = '''
Above all, My Lord, we want equal political rights, because without them our disabilities will be permanent.

I know this sounds revolutionary to the whites in this country, because the majority of voters will be Africans.

This makes the white man fear democracy.

But this fear cannot be allowed to stand in the way of the only solution which will guarantee racial harmony and freedom for all.

It is not true that the enfranchisement of all will result in racial domination.

Political division, based on colour, is entirely artificial and, when it disappears, so will the domination of one colour group by another.

The ANC has spent half a century fighting against racialism.

When it triumphs as it certainly must, it will not change that policy.

This then is what the ANC is fighting.

Our struggle is a truly national one.

It is a struggle of the African people, inspired by our own suffering and our own experience.

It is a struggle for the right to live.

During my lifetime I have dedicated my life to this struggle of the African people.

I have fought against white domination, and I have fought against black domination.

I have cherished the ideal of a democratic and free society in which all persons will live together in harmony and with equal opportunities.

It is an ideal for which I hope to live for and to see realised.

But, My Lord, if it needs be, it is an ideal for which I am prepared to die.
'''
````

````python
stopwords.words('english')
````

````python
stopwords.words('bengali')
````

````python
from nltk.stem import PorterStemmer
````

````python
porter_stemmer = PorterStemmer()
````

````python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
````

````python
sentences = sent_tokenize(paragraph)
````

````python
sentences
````

````python
type(sentences)
````

````python
len(sentences)
````

````python
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    print(words)
````

````python
# Remove stopwords and apply porter stemming
# Tokenize each sentence into words, remove stopwords and apply Porter stemming, then join the words back into a sentence.
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [porter_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
````

````python
sentences
````

````python
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer('english')
````

````python
# Remove stopwords and apply snowball stemming
# Tokenize each sentence into words, remove stopwords and apply Porter stemming, then join the words back into a sentence.
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [snowball_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
````

````python
sentences
````

````python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
````

````python
# Remove stopwords and apply lemmatizer
# Tokenize each sentence into words, remove stopwords and apply Porter stemming, then join the words back into a sentence.
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word, pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)
````

````python
sentences
````

````python
# Porter Stemmer = Fast and simple, but can produce rough/non-word stems
# Snowball Stemmer = Fast and more refined than Porter, but can still produce non-word stems
# Lemmatizer = Slower but produces meaningful dictionary/base words
````
