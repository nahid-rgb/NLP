# Named Entity Recognition (NER)


Named Entity Recognition (NER) is an NLP technique that identifies important real-world entities in text and assigns them a category.

For example:

"Nelson Mandela was born in South Africa in 1918."

- Nelson Mandela → PERSON
- South Africa   → GPE / LOCATION
- 1918           → DATE

In simple words:

NER finds names of people, places, organizations, dates, money, and other important entities in text.

#### Common Named Entities

| Entity       | Meaning              | Example               |
| ------------ | -------------------- | --------------------- |
| **PERSON**   | Person's name        | `Nelson Mandela`      |
| **GPE**      | Country, city, state | `Bangladesh`, `Dhaka` |
| **ORG**      | Organization/company | `Google`, `UN`        |
| **DATE**     | Date/year            | `1918`, `August 2026` |
| **MONEY**    | Money amount         | `$500`                |
| **TIME**     | Time                 | `10:30 AM`            |
| **LOCATION** | Physical location    | `Mount Everest`       |
| **PRODUCT**  | Product              | `iPhone`              |


```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
from nltk import word_tokenize, pos_tag, ne_chunk
```

```python
words = word_tokenize(sentence)
words
```

```python
tagged  = pos_tag(words)
tagged 
```

```python
import nltk 
nltk.download('punkt_tab')                    # Tokenization
nltk.download('stopwords')                    # Stopwords
nltk.download('averaged_perceptron_tagger_eng') # POS tagging
nltk.download('maxent_ne_chunker_tab')        # NER
nltk.download('words')                        # NER vocabulary
```

```python
ne_chunk(tagged)
```
