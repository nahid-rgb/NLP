# Parts of Speech (POS) Tagging



POS Tagging is the process of assigning a grammatical category to each word in a sentence.

In simple words, POS tagging tells us what role each word plays in a sentence.

Example:

"The beautiful girl runs quickly."

- The → Determiner
- beautiful → Adjective
- girl → Noun
- runs → Verb
- quickly → Adverb

#### Common Parts of Speech

| POS                  | Meaning                    | Example                |
| -------------------- | -------------------------- | ---------------------- |
| **Noun (NN)**        | Person, place, thing, idea | `boy`, `Dhaka`, `book` |
| **Verb (VB)**        | Action/state               | `run`, `eat`, `is`     |
| **Adjective (JJ)**   | Describes a noun           | `beautiful`, `big`     |
| **Adverb (RB)**      | Describes a verb/adjective | `quickly`, `very`      |
| **Pronoun (PRP)**    | Replaces a noun            | `he`, `she`, `it`      |
| **Preposition (IN)** | Shows relationship         | `in`, `on`, `at`       |
| **Conjunction (CC)** | Connects words/phrases     | `and`, `but`, `or`     |
| **Determiner (DT)**  | Specifies a noun           | `the`, `a`, `an`       |


```python
paragraph = '''
Above all, My Lord, we want equal political rights, because without them our disabilities will be permanent.

I know this sounds revolutionary to the whites in this country, because the majority of voters will be Africans.

This makes the white man fear democracy.

But this fear cannot be allowed to stand in the way of the only solution which will guarantee racial harmony and freedom for all.

It is not true that the enfranchisement of all will result in racial domination.

Political division, based on colour, is entirely artificial and, when it disappears, so will the domination of one colour group by another.

The ANC has spent half a century fighting against racialism.

When it triumphs as it certainly must, it will not change that policy.

This then is what the ANC is fighting.

Our struggle is a truly national one.

It is a struggle of the African people, inspired by our own suffering and our own experience.

It is a struggle for the right to live.

During my lifetime I have dedicated my life to this struggle of the African people.

I have fought against white domination, and I have fought against black domination.

I have cherished the ideal of a democratic and free society in which all persons will live together in harmony and with equal opportunities.

It is an ideal for which I hope to live for and to see realised.

But, My Lord, if it needs be, it is an ideal for which I am prepared to die.
'''
```

```python
from nltk.tokenize import sent_tokenize
```

```python
sentences = sent_tokenize(paragraph)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.tokenize import word_tokenize
```

```python
from nltk import pos_tag
```

```python
# Tokenize sentences, remove stopwords, and assign POS tags to the remaining words.
stop_words = set(stopwords.words('english'))

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [word for word in words if word not in stop_words]

    # Assign POS tags
    pos_tags = pos_tag(words)
    print(pos_tags)
```
