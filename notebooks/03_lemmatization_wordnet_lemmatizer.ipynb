# Lemmatization



#### What is Lemmatization?
Lemmatization is a text preprocessing technique that converts a word into its proper dictionary/base form, it uses **grammar rules and a dictionary** to find the actual **root/base word (called a "lemma")**.
Unlike stemming, lemmatization tries to produce a real English word.

In simple words: **Lemmatization gives the real dictionary word, not a chopped-up guess.**

#### Stemming vs Lemmatization (quick reminder)

| Word | Stemming | Lemmatization |
|---|---|---|
| "studies" | studi ❌ | study ✅ |
| "running" | run ✅ | run ✅ |
| "better" | better ❌ (no change) | good ✅ |
| "went" | went ❌ (no change) | go ✅ |

Lemmatization understands meaning and grammar, so it can even convert "better" → "good" or "went" → "go" — something stemming can never do.

#### WordNet Lemmatizer
WordNetLemmatizer is an NLTK tool that uses WordNet, a large English dictionary/database, to find the base form of a word.

```python
import nltk
print(nltk.__file__)
print(nltk.__version__)
```

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')   # needed for wordnet to work properly in newer NLTK versions
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going")     # Output: going , by default treated as noun, no change
```

```python
lemmatizer.lemmatize("going", pos='v') # Output: go treated as verb
```

```python
# The "pos" Parameter in WordNetLemmatizer
# "pos" stands for Part of Speech — it tells the lemmatizer what role the word plays in a sentence
# (noun, verb, adjective, etc.), so it can find the correct root word.

# Why does `pos` matter so much?
# The same word can have different meanings depending on whether it's a noun, verb, or adjective. 
# WordNet stores different lemma groups for each. If you don't tell it the correct `pos`, it defaults to 
# noun — and gives wrong or unchanged results.
```

| `pos` | Meaning   | Example             |
| ----- | --------- | ------------------- |
| `'n'` | Noun      | `cars → car`        |
| `'v'` | Verb      | `going → go`        |
| `'a'` | Adjective | `better → good`     |
| `'r'` | Adverb    | `quickly → quickly` |


```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos='n'))
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos='a'))
```
