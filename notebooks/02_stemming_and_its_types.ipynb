# Stemming
Stemming is a text preprocessing technique that reduces words to their root/base form by removing prefixes or suffixes.

It may not always produce a proper English word.

```
playing  → play
played   → play
plays    → play
```
Another example:

```
studies  → studi
studying → studi
```
Here, "studi" is not a proper English word, but stemming reduces similar words to a common form.

````python

words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
````

#### PorterStemmer

````python
from nltk.stem import PorterStemmer
````

````python
port_stemmer = PorterStemmer()
````

````python
for word in words:
    print(word+"---->"+port_stemmer.stem(word))
````

````python
port_stemmer.stem('congratulations')
````

````python
port_stemmer.stem("sitting")
````

````python
# Porter Stemmer is fast and useful, but it works mainly by cutting words rather than understanding their meaning.
# It chops off too much of the word, turning it into something with a completely different or wrong meaning.
````

#### RegexpStemmer Class



`RegexpStemmer` is a stemmer from NLTK that lets **you** define your own rule (using regex) for how words should be stemmed — instead of relying on a fixed algorithm like Porter Stemmer.

In simple words: **you tell it which pattern/ending to remove, and it removes that pattern from the words.**


````python
from nltk.stem import RegexpStemmer
````

````python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
````

````python
reg_stemmer.stem('eating')
````

````python
reg_stemmer.stem('ingeating')
````

#### Snowball Stemmer



Snowball Stemmer (also called **"Porter2"**) is an **improved and more accurate version of Porter Stemmer**, created by the same person (Martin Porter). It's more aggressive-but-smarter with its rules and supports **multiple languages**, not just English.

In simple words: **it's Porter Stemmer's upgraded version — better logic, cleaner results, and works for many languages.**

````python
from nltk.stem import SnowballStemmer
````

````python
snowball_stemmer = SnowballStemmer('english')
````

````python
for word in words:
    print(word+"---->"+snowball_stemmer.stem(word))
````

````python
snowball_stemmer.stem("fairly"),snowball_stemmer.stem("sportingly")
````

#### Porter vs Regexp vs Snowball — Simple Strengths & Weaknesses

### 1. Porter Stemmer
- ✅ Strength: Simple, fast, easy to use
- ❌ Weakness: Often cuts words wrong, gives non-real words
  - "studies" → "studi"

### 2. RegexpStemmer
- ✅ Strength: You decide the rule yourself, full control
- ❌ Weakness: No brain — removes pattern even when wrong
  - "king" → "k"

### 3. Snowball Stemmer
- ✅ Strength: Better/cleaner than Porter, works for many languages
- ❌ Weakness: Still not perfect, still no real dictionary words sometimes
  - "studies" → "studi"

### One-line memory trick
- **Porter** = old & basic
- **Regexp** = manual & risky
- **Snowball** = upgraded Porter
