# Word2Vec
Word2Vec is a word embedding technique that converts words into numerical vectors based on their surrounding context.

It helps capture relationships between words, where words with similar meanings or contexts can have similar vector representations.

### Training Word2Vec from Scratch: CBOW and Skip-Gram

#### CBOW (Continuous Bag of Words)

```python
from gensim.models import Word2Vec
```

```python
corpus = [
    "go until jurong point crazy available",
    "ok lar joking wif u oni",
    "free entry wkly comp win cup final"
]
```

```python
tokenized_corpus = [sentence.split() for sentence in corpus]
tokenized_corpus
```

```python
print(tokenized_corpus[0])
```

##### Train CBOW from scratch

```python
from gensim.models import word2vec

# Train CBOW Word2Vec model
cbow_model = Word2Vec(
    sentences=tokenized_corpus, # Training sentences
    vector_size=100,            # Each word will have 100 numbers
    window=3,                   # Context window size
    min_count=1,                # Ignore words appearing less than 1 time
    sg=0,                       # 0 = CBOW, 1 = Skip-Gram
    epochs=10                   # Train the model 10 times=10 
)
```

```python
# Check all words in the vocabulary
print(cbow_model.wv.index_to_key)
```

```python
cbow_model.wv['free']
```

```python
cbow_model.wv['nlp']
```

##### Train Skip-Gram  from scratch

```python
from gensim.models import Word2Vec

# Train Skip-Gram Word2Vec model
skipgram_model = Word2Vec(
    sentences=tokenized_corpus,     # Training sentences
    vector_size=100,                # Each word gets 100 numbers
    window=3,                       # Context window size
    min_count=1,                    # Keep words appearing at least 1 time
    sg=1,                           # 0 = CBOW, 1 = Skip-Gram
    epochs=10                       # Train the model 10 times
)
```

```python
# Get the vector for a word
skipgram_model.wv['free']
```

```python
# Find words similar to "free"
skipgram_model.wv.most_similar('free')
```

```python
# Compare two words
skipgram_model.wv.similarity('free', 'entry')
```

```python
cbow_model.wv.similarity('free', 'entry')
```

```python
print(len(cbow_model.wv['free']))
print(len(skipgram_model.wv['free']))
```
